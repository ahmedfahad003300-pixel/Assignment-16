```python
#7
df=df.dropna(subset=["YIELD_KG"])
df
```

```python
#8
df["AREA_ACRES"] = df["AREA_ACRES"].fillna(df["AREA_ACRES"].mean())
df
```

```python
#9
df["DISTRICT"] = df["DISTRICT"].str.strip().replace("", "unknown")
df
```

```python
#10
df["FERTILIZER_USED"] = df["FERTILIZER_USED"].str.strip().replace("", "NO")
df
```

```python
#11
df["YIELD_VALUE"].fillna (df["YIELD_VALUE"].mean())
df
```

```python
#12
df["HARVEST_DATE"] = pd.to_datetime(df["HARVEST_DATE"], errors="coerce")
df.dtypes
```

```python
#13
df["FARM_ID"] = df["FARM_ID"].str.replace("F", "")
df["FARM_ID"] = df["FARM_ID"].astype(int)
df
```

```python
#16
group_by_crop=df.groupby("CROP_TYPE")["YIELD_KG"].sum()
print(group_by_crop,"kg")
```

```python
#17
import matplotlib.pyplot as plt
import numpy as np

sum_by_crop=df.groupby("CROP_TYPE")["YIELD_KG"].sum()
unique_crop=df["CROP_TYPE"].unique()
plt.bar(unique_crop,sum_by_crop, color='green', width=0.5)
plt.xlabel("unique_crop")
plt.ylabel("sum_by_crop")
plt.title('Bar Plot')
plt.show()
```
